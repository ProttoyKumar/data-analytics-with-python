```python
# DataFrame 1: City and Temperature
df_temp = pd.DataFrame({
    'city': ['New York', 'Los Angeles', 'Chicago'],
    'temperature': [28, 31, 24]
})

# DataFrame 2: City and Wind Speed
df_wind = pd.DataFrame({
    'city': ['New York', 'Chicago'],
    'windspeed': [12, 18]
})
```

```python
pd.merge(df_temp,df_wind,on='city')
```

```python
pd.merge(df_temp,df_wind,how='outer')
```

```python
pd.merge(df_temp,df_wind,how='inner')
```

how='inner' ----> by default

```python
pd.merge(df_temp,df_wind,how='left')
```

```python
pd.merge(df_temp,df_wind,how='right')
```
